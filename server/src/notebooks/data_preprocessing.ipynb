```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('../data/data.csv')
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
(df == '?').sum()
```

```python
df['mindStatus'] = df['mindStatus'].astype('category')

label_mapping = dict(enumerate(df['mindStatus'].cat.categories))
label_mapping
```

```python
df['mindStatus'] = df['mindStatus'].cat.codes
df
```

```python
features = df.drop(columns=['mindStatus'], axis=1)
labels = df['mindStatus']
```

```python
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)
```

```python
processed_df = pd.DataFrame(features_scaled, columns=features.columns)
processed_df['mindStatus'] = labels
processed_df.head()
```

```python
processed_df.to_csv('../data/processed_data.csv', index=False)
print("Processed data saved to 'processed_training_data.csv'")
```
